```python
import pandas as pd
df = pd.read_csv('customer_shopping_behavior.csv')
df.head()
```

```python
df.describe()
```

```python
df['Customer ID'].duplicated().sum()
```

```python
df.info()
```

```python
df["Review Rating"]= df["Review Rating"].fillna(df.groupby("Category")["Review Rating"].transform("median"))
df["Review Rating"]
```

```python
df.isnull().sum()
```

```python
df.columns= df.columns.str.lower()
df.columns=df.columns.str.replace(" ","_")
df= df.rename(columns={"purchase_amount_(usd)":"purchase_amount"})
df.columns
```

```python
#creating a age_group column
df['age_group'] = pd.qcut(df["age"],q=4,labels=['Young Adult','Adult','Middel Aged','Senior'])
```

```python
df[['age','age_group']].head(10)
```

```python
df.groupby("frequency_of_purchases").agg({"customer_id":"count"})
```

```python
frequency_mapping={
    "Fortnightly":14,
    "Monthly":30,
    "Quarterly":90,
    "Weekly":7,
    "Annually":365,
    "Bi-Weekly":14,
    "Every 3 Months":90
}
df["purcahse_frequency_days"] = df["frequency_of_purchases"].map(frequency_mapping)
df[["frequency_of_purchases","purcahse_frequency_days"]].head(10)
```
